```python
import numpy as np
import random
```

```python
# Distance matrix (example)
dist = np.array([
    [0, 2, 2, 5, 7],
    [2, 0, 4, 8, 2],
    [2, 4, 0, 1, 3],
    [5, 8, 1, 0, 2],
    [7, 2, 3, 2, 0]
])
```

```python
n = len(dist)
pheromone = np.ones((n, n))
alpha, beta = 1, 2
evaporation = 0.5
Q = 100
```

```python
def route_length(route):
    return sum(dist[route[i]][route[i+1]] for i in range(len(route)-1)) + dist[route[-1]][route[0]]


def choose_next(city, visited):
    probs = []
    for j in range(n):
        if j not in visited:
            tau = pheromone[city][j] ** alpha
            eta = (1 / dist[city][j]) ** beta if dist[city][j] != 0 else 0
            probs.append((j, tau * eta))
    total = sum(p for _, p in probs)
    r = random.random()
    s = 0
    for j, p in probs:
        s += p / total
        if r <= s:
            return j


def aco(iterations=50, ants=10):
    best_route = None
    best_length = float('inf')

    for _ in range(iterations):
        all_routes = []
        for _ in range(ants):
            route = [random.randint(0, n-1)]
            while len(route) < n:
                route.append(choose_next(route[-1], route))
            all_routes.append(route)

            length = route_length(route)
            if length < best_length:
                best_length = length
                best_route = route

        pheromone[:] *= (1 - evaporation)

        for route in all_routes:
            length = route_length(route)
            for i in range(n-1):
                pheromone[route[i]][route[i+1]] += Q / length

    return best_route, best_length
```

```python
route, length = aco()
print("Best route:", route)
print("Shortest distance:", length)
```
